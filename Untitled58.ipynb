```python
# Cloud-Based Bus Pass System
# CodeAlpha - Task 3

routes = {
    "Chennai - Pondicherry": 150,
    "Chennai - Chengalpattu": 80,
    "Chennai - Mahabalipuram": 100,
    "Pondicherry - Chengalpattu": 120,
    "Chengalpattu - Mahabalipuram": 60
}

print("===== BUS ROUTES & FARES =====")

for route, fare in routes.items():
    print(f"{route} : ₹{fare}")

print("\nTotal Routes:", len(routes))
```

```python
# Step 2: Passenger Booking System

import uuid

name = input("Enter passenger name: ")
phone = input("Enter phone number: ")
route = input("Enter route: ")

if route in routes:
    fare = routes[route]

    ticket_id = "BUS-" + uuid.uuid4().hex[:8].upper()

    print("\n===== BOOKING CONFIRMED =====")
    print("Passenger Name :", name)
    print("Phone Number   :", phone)
    print("Route          :", route)
    print("Fare           : ₹", fare)
    print("Ticket ID      :", ticket_id)
    print("Status         : Confirmed")

else:
    print("Invalid route. Please select a valid route.")
```

```python
# Step 3: Store Booking Details in SQLite

import sqlite3

# Create database
connection = sqlite3.connect("bus_pass.db")
cursor = connection.cursor()

# Create booking table
cursor.execute("""
CREATE TABLE IF NOT EXISTS bookings (
    ticket_id TEXT PRIMARY KEY,
    passenger_name TEXT,
    phone TEXT,
    route TEXT,
    fare INTEGER,
    status TEXT
)
""")

# Save the booking
cursor.execute("""
INSERT INTO bookings
(ticket_id, passenger_name, phone, route, fare, status)
VALUES (?, ?, ?, ?, ?, ?)
""", (ticket_id, name, phone, route, fare, "Confirmed"))

connection.commit()

print("Booking saved successfully!")
print("Ticket ID:", ticket_id)

connection.close()
```

```python
# Step 4: Display All Bookings

import sqlite3

connection = sqlite3.connect("bus_pass.db")
cursor = connection.cursor()

cursor.execute("SELECT * FROM bookings")
bookings = cursor.fetchall()

print("===== ALL BUS BOOKINGS =====")

for booking in bookings:
    print("\nTicket ID      :", booking[0])
    print("Passenger Name :", booking[1])
    print("Phone Number   :", booking[2])
    print("Route          :", booking[3])
    print("Fare           : ₹", booking[4])
    print("Status         :", booking[5])

connection.close()
```

```python
# Step 5: Ticket Verification

import sqlite3

ticket = input("Enter Ticket ID: ")

connection = sqlite3.connect("bus_pass.db")
cursor = connection.cursor()

cursor.execute(
    "SELECT * FROM bookings WHERE ticket_id = ?",
    (ticket,)
)

booking = cursor.fetchone()

if booking:
    print("\n===== VALID TICKET =====")
    print("Ticket ID      :", booking[0])
    print("Passenger Name :", booking[1])
    print("Phone Number   :", booking[2])
    print("Route          :", booking[3])
    print("Fare           : ₹", booking[4])
    print("Status         :", booking[5])
else:
    print("\nInvalid Ticket ID!")
    print("Ticket not found.")

connection.close()
```

```python
# Step 6: Input Validation

name = input("Enter passenger name: ").strip()
phone = input("Enter phone number: ").strip()
route = input("Enter route: ").strip()

# Validate passenger name
if name == "":
    print("Error: Passenger name cannot be empty.")

# Validate phone number
elif not phone.isdigit() or len(phone) != 10:
    print("Error: Enter a valid 10-digit phone number.")

# Validate route
elif route not in routes:
    print("Error: Invalid route selected.")

else:
    fare = routes[route]

    print("\n===== VALID BOOKING DETAILS =====")
    print("Passenger Name :", name)
    print("Phone Number   :", phone)
    print("Route          :", route)
    print("Fare           : ₹", fare)
    print("All details are valid!")
```

```python
# Step 7: Duplicate Booking Prevention

import sqlite3

connection = sqlite3.connect("bus_pass.db")
cursor = connection.cursor()

# Check whether the Ticket ID already exists
cursor.execute(
    "SELECT ticket_id FROM bookings WHERE ticket_id = ?",
    (ticket_id,)
)

existing_ticket = cursor.fetchone()

if existing_ticket:
    print("Booking already exists!")
    print("Ticket ID:", ticket_id)
else:
    cursor.execute("""
    INSERT INTO bookings
    (ticket_id, passenger_name, phone, route, fare, status)
    VALUES (?, ?, ?, ?, ?, ?)
    """, (ticket_id, name, phone, route, fare, "Confirmed"))

    connection.commit()
    print("New booking saved successfully!")
    print("Ticket ID:", ticket_id)

connection.close()
```

```python
# Step 8: Search Booking by Passenger Name

import sqlite3

search_name = input("Enter passenger name: ").strip().lower()

connection = sqlite3.connect("bus_pass.db")
cursor = connection.cursor()

cursor.execute("""
SELECT * FROM bookings
WHERE LOWER(REPLACE(passenger_name, ' ', '')) = ?
""", (search_name.replace(" ", ""),))

results = cursor.fetchall()

if results:
    print("\n===== BOOKING DETAILS =====")

    for booking in results:
        print("Ticket ID      :", booking[0])
        print("Passenger Name :", booking[1])
        print("Phone Number   :", booking[2])
        print("Route          :", booking[3])
        print("Fare           : ₹", booking[4])
        print("Status         :", booking[5])
        print("-----------------------------")
else:
    print("No booking found for this passenger.")

connection.close()
```

```python
# Step 9: Cancel a Booking

import sqlite3

ticket_id = input("Enter Ticket ID to cancel: ").strip()

connection = sqlite3.connect("bus_pass.db")
cursor = connection.cursor()

cursor.execute(
    "SELECT status FROM bookings WHERE ticket_id = ?",
    (ticket_id,)
)

booking = cursor.fetchone()

if booking is None:
    print("Ticket not found.")

elif booking[0] == "Cancelled":
    print("This ticket is already cancelled.")

else:
    cursor.execute(
        "UPDATE bookings SET status = 'Cancelled' WHERE ticket_id = ?",
        (ticket_id,)
    )

    connection.commit()

    print("\n===== BOOKING CANCELLED =====")
    print("Ticket ID:", ticket_id)
    print("Status: Cancelled")

connection.close()
```

```python
# Step 12: Booking Statistics

import sqlite3

connection = sqlite3.connect("bus_pass.db")
cursor = connection.cursor()

# Total bookings
cursor.execute("SELECT COUNT(*) FROM bookings")
total_bookings = cursor.fetchone()[0]

# Confirmed bookings
cursor.execute("SELECT COUNT(*) FROM bookings WHERE status = 'Confirmed'")
confirmed_bookings = cursor.fetchone()[0]

# Cancelled bookings
cursor.execute("SELECT COUNT(*) FROM bookings WHERE status = 'Cancelled'")
cancelled_bookings = cursor.fetchone()[0]

# Total fare from confirmed bookings
cursor.execute("""
SELECT COALESCE(SUM(fare), 0)
FROM bookings
WHERE status = 'Confirmed'
""")
total_fare = cursor.fetchone()[0]

print("===== BOOKING STATISTICS =====")
print("Total Bookings     :", total_bookings)
print("Confirmed Bookings :", confirmed_bookings)
print("Cancelled Bookings :", cancelled_bookings)
print("Total Fare         : ₹", total_fare)

connection.close()
```

```python
%%writefile app.py

import streamlit as st
import sqlite3
import uuid

# Bus routes and fares
routes = {
    "Chennai - Pondicherry": 150,
    "Chennai - Chengalpattu": 80,
    "Chennai - Mahabalipuram": 100,
    "Pondicherry - Chengalpattu": 120,
    "Chengalpattu - Mahabalipuram": 60
}

# Create database
connection = sqlite3.connect("bus_pass.db")
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS bookings (
    ticket_id TEXT PRIMARY KEY,
    passenger_name TEXT,
    phone TEXT,
    route TEXT,
    fare INTEGER,
    status TEXT
)
""")

connection.commit()
connection.close()

# Page title
st.title("🚌 Cloud-Based Bus Pass System")
st.write("Book your bus pass easily and securely.")

# Passenger details
name = st.text_input("Passenger Name")
phone = st.text_input("Phone Number")

route = st.selectbox(
    "Select Route",
    list(routes.keys())
)

fare = routes[route]

st.write("**Fare:** ₹", fare)

# Booking button
if st.button("Book Bus Pass"):

    if name.strip() == "":
        st.error("Please enter passenger name.")

    elif not phone.isdigit() or len(phone) != 10:
        st.error("Please enter a valid 10-digit phone number.")

    else:
        ticket_id = "BUS-" + uuid.uuid4().hex[:8].upper()

        connection = sqlite3.connect("bus_pass.db")
        cursor = connection.cursor()

        cursor.execute("""
        INSERT INTO bookings
        (ticket_id, passenger_name, phone, route, fare, status)
        VALUES (?, ?, ?, ?, ?, ?)
        """, (ticket_id, name, phone, route, fare, "Confirmed"))

        connection.commit()
        connection.close()

        st.success("Booking Confirmed!")

        st.write("### 🎫 Ticket Details")
        st.write("**Ticket ID:**", ticket_id)
        st.write("**Passenger:**", name)
        st.write("**Route:**", route)
        st.write("**Fare:** ₹", fare)
        st.write("**Status:** Confirmed")
```
